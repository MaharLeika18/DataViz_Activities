```python
# Import modules
import pandas as pd
```

```python
# Load raw dataset
df = pd.read_csv('juanmart_raw_sales.csv')
```

```python
# Identify data anomalies
display(df)

# The formatting of values in column order_date is inconsistent and alternates between - and / as the value separator. The 
# letter case of the strings in region is also inconsistent with some entries having the same string but in different casing.
```

```python
df.dtypes

# Column datatypes are as expected.
```

```python
df.info()

# RangeIndex reveals that there are 12 entries or rows in the dataset. Columns cust_name and amount_paid have 10 non-null
# entries or, in other words, 2 null entries. 
```

```python
df.describe()

# describe() reveals the statistics of the numerical columns transaction_id and amount_paid. There are only 10 entries in
# amount_paid out of the expected 12. There are no unexpected negative values in the amount_paid column.
```

```python
df.isnull().sum()

# This confirms that there are a total of 4 null values in the dataset; two in column cust_name and two in column amount_paid
```

```python
df.value_counts()

# Reveals that there are two entries that have duplicates in the dataset namely transaction 1001 and 1005.
```

```python
test = df
test["order_date"] = pd.to_datetime(df["order_date"], errors="coerce")    # convert values into datetime if valid
display(test)

# Rows 1, 3, 6, 8, and 11 have invalid value formats and could not be converted into datetime format.
```

```python
# Data cleaning
df['order_date'] = df['order_date'].str.replace("-", "/", regex=False)
df['order_date'] = pd.to_datetime(df['order_date'], format='%Y/%m/%d',errors='coerce')
df['region'] = df['region'].str.lower()
df.drop_duplicates(inplace=True)
df['amount_paid'] = df['amount_paid'].fillna(0)
df['cust_name'] = df['cust_name'].fillna('Unknown')

display(df)
```
